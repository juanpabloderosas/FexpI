***d. Una de las maneras de tratar con este tipo de datos, es hacer un promedio. Escriba la definición de
promedio, calcule e informe. En términos más formales, al promedio le decimos media muestral, y en el
caso en que sea una media muestral de una colección de valores $g_i$ , escribiremos la media muestral con
una barra arriba, $\bar{g}$.***

Recordamos que la ***media muestral***, conocido habitualmente como *promedio* de un número de una colección de $n$ elementos, $g_1,g_2,g_3,...,g_n$, está dada por:

$$\bar{g} = \frac{g_1+g_2+g_3+...+g_n}{n} = \frac{1}{n}\sum_{i=1}^{n}g_i$$

hacerlo con calculadora para 20 valores puede ser incómodo. Para 4000 valores es un trabajo inadmisible.

```r
#g_acel es la media (mean, en inglés) de los datos g
g_acel <- mean(datos$g)
cat(bold$blue("g_media_muestral = ", g_acel, " m/s^2"))
```

**e) Dibuje un histograma de las medidas. ¿Por qué la media muestral serı́a representativa?**

Abajo está el código para hacer un histograma, es de la librería ggplot2.

Tiene bastantes cosas, lo que es importante destacar es que cada línea define diferentes cosas, desde el histograma, pasando por las líneas verticales indicando $\bar{g}$ y $g_{RAGA}$ , además de títulos de gráficos, colores, etc.

Pueden probar cambiando algo, por gusto por ahora, luego veremos que los gráficos en ciencias tienen su *manera estándard* de armarse.

```r
glimpio <- datos$g

histo <- ggplot() + #crea objeto ggplot
  #dibuja histograma
  geom_histogram(aes(glimpio),bins = nclass.FD(glimpio), colour = "Black", alpha = 0.4, lwd = 0.2, show.legend = F) + #la capa del histograma
  # líneas horizontales
  geom_vline(aes(xintercept = g_acel, colour = "g acel" ), lwd = 2 ) + 
  geom_vline(aes(xintercept = g_RAGA, colour = "g RAGA" ), lwd = 2 ) + 
  # color para identificar líneas
  scale_colour_manual(values = c("g acel" = "Red3", "g RAGA" = "DeepSkyBlue4"),
                      labels = list(TeX("$\\; \\bar{g}_{acel}$"),TeX("$\\;\\;\\; \\bar{g}_{RAGA}$")),name = NULL) +
  
#títulos gráfico y ejes
  ylab("Frecuencia") + xlab(TeX("$g \\; (m \\, s^{-2} ) $")) +
  ggtitle("Medidas del Acelerómetro (Franco)") + 
  scale_y_continuous(expand = expansion(mult = c(0,0.1)) ) + 
  #detalles de la leyenda para leer el gráfico
    temajuan + 
  theme(legend.position = c(0.15,0.8), legend.direction = "horizontal", legend.key.height = unit(48,"pt"),
        legend.key.width = unit(72,"pt"),  legend.text =  element_text(hjust = 12),
        legend.background = element_blank()) 


fig(width = 24,14)
```

***f ) Calcule la desviación estándar de la muestra, s. Defínala, calculela e informe el valor. ¿Por qué s (la desviación estándar de la muestra) sería una medida de dispersión de la medida?***

$$s_g =\frac{1}{\sqrt{n-1}}\sqrt{\Big((\,g_1 - \bar{g} \,)^2 + (\,g_2 - \bar{g} \,)^2 + ...+(\,g_n - \bar{g} \,)^2 \Big)} =\sum_{i = 1}^{n}\sqrt{\frac{(\,g_i - \bar{g} \,)^2}{n-1}}$$

La desviación estándar es una medida de cuán disperso está el conjunto de valores respecto de la media muestral de los datos. Para una distribución de valores $g_i$, los valores que se encuentren más lejos de $\bar{g}$ aportarán mucho al valor de $s_g$

```r
sg <- sd(datos$g) #la desviación estándar se calcula con la función sd(vector numérico)

cat(bold$blue("La desviación estándard g_Franco es \n\ns_g = ", sg, " m/s^2"))
```

la desviación estándar, además, tiene las unidades de la media muestral.

Es habitual resumir los resultados del histograma (bajo ciertas condiciones que veremos luego), expresando la media muestral +- la desviación estándar, es decir:

$$\boxed{\bar{g} \pm s_g = (9.8263 \;\pm \;0.0291) m s^{-2} \;}$$

en la desviación estándar no se usa, en general, el redondeo a una cifra significativa, pero la cantidad de cifras a usar depende de cosas que por ahora no veremos. Le dejamos cuatro cifras significativas y redondeamos $\bar{g}$ en el lugar decimal correspondiente.

```r
#vemos los nombres de archivos
setwd(dirDatos)
archivos <- list.files(); #todos los archivos
archivos <- archivos[grep(archivos, pattern = ".csv")] #sólo los que sean de extensión .csv
archivos

indice <- vector(mode = "numeric");
t      <- vector(mode = "numeric");
g      <- vector(mode = "numeric");
Nombre <- vector(mode = "character");

for(i in archivos){
g_individual <- read.csv(i, header = F, skip = 1, sep = ","); #cargamos i = "g_Agustina.csv", "g_Antonella.csv"..."g_Tadeo.csv"
    
g_individual <- g_individual[, -(2:4)];

colnames(g_individual) <- c("t", "g");    
    
nombre <- rep(gsub(x = i, c("g_|.csv"), replacement = ""), times = nrow(g_individual));

indice = c(indice,1:nrow(g_individual))
t <- c(t, g_individual$t)    
g <- c(g, g_individual$g)    
Nombre <- c(Nombre, nombre)
    


}

d_tutti <- data.frame(indice , t , g , Nombre)
head(d_tutti);
tail(d_tutti)
```

```r
g_media <- vector(mode = "numeric");
sd_g <-  vector(mode = "numeric");
Nombre <- unique(d_tutti$Nombre, );


for(i in 1:length(Nombre)){
g_individual <- d_tutti[d_tutti$Nombre == Nombre[i], ]; 

g_media[i] <- mean(g_individual$g);
sd_g[i] <- sd(g_individual$g)    
}

resumen <- data.frame(Nombre,g_media, sd_g)
```

Las barritas de incerteza indicando la desviación estándar, en este caso tienen sentido sólo si las medidas son ***estables***. Es por esto que necesitamos ver un gráfico de dispersión antes de hace un histograma, o tener los ojos muy afinados viendo todos los histogramas juntos.


De hecho, se puede hacer un gráfico de ***violín***, que son como histogramas espejados, y mirar qué pasa:

```r
gviolin <- ggplot() + 
geom_violin(data = d_tutti,aes(x = Nombre,y = g), fill = "Deepskyblue4", alpha = 0.3) + 
geom_pointrange(data = resumen, aes(x = Nombre,y = g_media, ymin = g_media - sd_g, ymax = g_media + sd_g )) + 
geom_hline(aes(yintercept = g_RAGA, lty = "g_Raga"), lwd = 1) +  
xlab(NULL) + 
ylab(TeX("$g (m \\, s^{-2})$")) + 
temajuan +
theme(legend.direction = "horizontal",
      legend.position = "top",
      legend.title = element_blank())

gviolin
```

```r

gviolin <- ggplot() + 
geom_violin(data = dnuevo,aes(x = Nombre,y = g), fill = "Deepskyblue4", alpha = 0.3) + 
#geom_pointrange(data = resumen, aes(x = Nombre,y = g_media, ymin = g_media - sd_g, ymax = g_media + sd_g )) + 
geom_hline(aes(yintercept = g_RAGA, lty = "g_Raga"), lwd = 1) +  
ylab(TeX("$g (m \\, s^{-2})$")) + xlab(NULL) +
temajuan +
theme(legend.direction = "horizontal",
      legend.position = "top",
      legend.title = element_blank())

gviolin
```
